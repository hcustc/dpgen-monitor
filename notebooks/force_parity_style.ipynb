```
import numpy as np
import os
import argparse
import matplotlib.pyplot as plt 
import matplotlib
import matplotlib.colors
from mpl_toolkits.axes_grid1 import make_axes_locatable
from scipy.interpolate import interpn

s=10

f=np.load("drive/My Drive/Undergraduate_Work/dodecane/dodecane_new_sel_f.npy")

#f_error=f[:,0]-f[:,1]
a=f[:, 0].ravel()
b=f[:, 1].ravel()
#a=f[:,0].ravel()
#b=f[:,1].ravel()
print("f_mean", np.mean(np.sqrt(np.sum(np.square(f[:,:3]), axis=1))))
print("f_mean", np.mean(np.sqrt(np.sum(np.square(f[:,3:6]), axis=1))))
f_error = a-b 
f_mae=np.mean(np.abs(f_error))
f_rmse=np.sqrt(np.mean(np.square(f_error)))
#e=np.concatenate([np.loadtxt(os.path.join(folder,filename),ndmin=2) for filename in os.listdir(folder) if filename.endswith('e.out')])


#e_rmse=np.sqrt(np.mean(np.square(e[:,:1]-e[:,1:2])))
#print("No of energy: ",e.shape[0],"RMSE: ",e_rmse)
print("No of energies: ",f.shape[0],"MAE",f_mae,"RMSE: ",f_rmse)

#xy = np.vstack([a,b])
#z = gaussian_kde(xy)(xy)
data , x_e, y_e = np.histogram2d( a, b, bins = 10000)

z = interpn( ( 0.5*(x_e[1:] + x_e[:-1]) , 0.5*(y_e[1:]+y_e[:-1]) ) , data , np.vstack([a,b]).T , method = "splinef2d", bounds_error = False )
z[z<0]=0
#z = np.log10(z+1)
#plt.scatter(f[:,:3].ravel(),f[:,3:6].ravel(),s=s)
font = {#'family' : 'normal',
    'size'   : 60} 

matplotlib.rc('font', **font)                                                                                                                                                                                                                                                                                             
idx = z.argsort()

a, b, z = a[idx], b[idx], z[idx]
fig=plt.figure(figsize=(32,32))
ax2 = plt.axes()                                                                                                                                                                                                                                                                                                          

im=ax2.scatter(a,b,s=s,c=z, cmap='rainbow', edgecolor='', zorder=30, norm=matplotlib.colors.LogNorm(vmin=1))

divider = make_axes_locatable(ax2)
cax = divider.append_axes("right", size="5%", pad=0.05)
fig.colorbar(im,ax=ax2, cax=cax)
ax2.set_aspect('equal')

lims = [
    np.min([ax2.get_xlim(), ax2.get_ylim()]),  # min of both axes
    #0,
    np.max([ax2.get_xlim(), ax2.get_ylim()]),  # max of both axes
]
lims=[-101,101]
#lims[1]=0
#ax2.set_xlim(lims)
ax2.set_xlim(lims)
ax2.set_ylim(lims)
#ax2.set_xlabel('DFT energies (kcal/mol)')
#ax2.set_ylabel('DL energies (kcal/mol)')
ax2.set_xlabel('DFT forces (eV/Å)')
ax2.set_ylabel('DP forces (eV/Å)')
#ax2.set_ylim(lims)
ax2.plot(lims, lims, 'k-', color='blue',linestyle="--", linewidth=10, dashes=(2, 2), zorder=20)
#alpha=0.75
ax2.set_xticks(np.arange(-100,100+1,50))
ax2.set_yticks(np.arange(-100,100+1,50))


ax3 = fig.add_axes([0.3, 0.6, 0.2, 0.2])
ax3.scatter(a,b,s=s, c=z, cmap='rainbow', edgecolor='', zorder=30, norm=matplotlib.colors.LogNorm(vmin=1))
ax3.plot([-11,11],[-11,11],'k-', alpha=0.75, color='r')
ax3.set_xlim([-11,11])
ax3.set_ylim([-11,11])

ax2.text(lims[0]+(lims[1]-lims[0])*0.3,lims[0]+(lims[1]-lims[0])*0.1,f"MAE={f_mae:.2f} eV/Å\nRMSE={f_rmse:.2f} eV/Å")
#ax2.text(lims[0]+(lims[1]-lims[0])*0.3,lims[0]+(lims[1]-lims[0])*0.1,f"MAE={f_mae:.2f} kcal/mol\nRMSE={f_rmse:.2f} kcal/mol")

[i.set_linewidth(5) for i in ax3.spines.values()]
[i.set_linewidth(5) for i in ax2.spines.values()]
ax2.xaxis.set_tick_params(width=10,size=25)
ax2.yaxis.set_tick_params(width=10,size=25)
#ax2.xaxis.set_ticks_position('both') 
#ax2.xaxis.set_tick_params(direction='in')
#ax2.xaxis.set_tick_params(direction='in')
#ax2.yaxis.set_ticks_position('both') 
#ax2.yaxis.set_tick_params(direction='in')
#ax2.yaxis.set_tick_params(direction='in')
[i.set_linewidth(10) for i in ax2.spines.values()]

fig.set_tight_layout(True)
ax3.xaxis.set_tick_params(width=5,size=20)
ax3.yaxis.set_tick_params(width=5,size=20)
fig.show()
```
